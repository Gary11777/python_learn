```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://dimgent.com'
response = requests.get(url)
```

```python
html_content = response.text
```

```python
soup = BeautifulSoup(html_content, 'html.parser')
```

```python
print(html_content[:500])
```

```python
links = soup.find_all('a')
```

```python
links
```

```python
for link in links:
    print(link.text);
```

```python
p = soup.find_all('p')
p
```

```python
for x in p:
    print(x.text);
```
